```python
import time

# Tạo dữ liệu giả lập: Tập A (1000 điểm), Tập B (1500 điểm) trong không gian 2D
np.random.seed(42)
A = np.random.rand(1000, 2)
B = np.random.rand(1500, 2)

# --- 1. Đo thời gian bằng Broadcasting ---
start_time = time.time()
diff = A[:, np.newaxis, :] - B[np.newaxis, :, :]
dist_broadcasting = np.sqrt(np.sum(diff**2, axis=-1))
time_broadcasting = time.time() - start_time

# --- 2. Đo thời gian bằng vòng lặp for ---
start_time = time.time()
N, M = len(A), len(B)
dist_loop = np.zeros((N, M))
for i in range(N):
    for j in range(M):
        dist_loop[i, j] = np.linalg.norm(A[i] - B[j])
time_loop = time.time() - start_time

print(f"Thời gian Broadcasting: {time_broadcasting:.5f} giây")
print(f"Thời gian dùng vòng lặp: {time_loop:.5f} giây")
print(f"Broadcasting nhanh gấp: {time_loop / time_broadcasting:.1f} lần")
```

```python
A = np.array([[1, 2], [3, 4]])
b = np.array([5, 6])
```

```python
# Nhân ma trận với ma trận
C = A @ A  # hoặc np.dot(A, A)

# Nhân ma trận với véc-tơ
result = A @ b

# Chuyển vị ma trận
A_transpose = A.T
```

```python
# 1. Tính định thức ma trận (Determinant)
det_A = np.linalg.det(A)

# 2. Tìm ma trận nghịch đảo (Inverse)
inv_A = np.linalg.inv(A)

# 3. Giải hệ phương trình tuyến tính Ax = b
# Ví dụ giải hệ phương trình: x + 2y = 5 và 3x + 4y = 6
x = np.linalg.solve(A, b)

# 4. Tìm trị riêng (Eigenvalues) và véc-tơ riêng (Eigenvectors)
eigenvalues, eigenvectors = np.linalg.eig(A)

# 5. Phân rã giá trị kỳ dị (SVD - Singular Value Decomposition)
U, S, Vt = np.linalg.svd(A)
```
