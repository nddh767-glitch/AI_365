```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
rng = np.random.default_rng(42)
x = np.linspace(0, 10, 100)
y = 3 * x + 2 + rng.normal(0, 1, size = 100)
```

```python
A = np.column_stack([x, np.ones_like(x)])
```

```python
theta_gd = np.zeros(2)
learning_rate = 0.01
n = len(y)

for step in range(5000):
    error = A @ theta_gd - y;
    gradient = (2 / n) * (A.T @ error)
    theta_gd -= learning_rate * gradient
```

```python
theta_ne = np.linalg.solve(A.T @ A, A.T @ y)
```

```python
model = LinearRegression()
model.fit(x.reshape(-1, 1), y)

theta_sklearn = np.array([
    model.coef_[0],
    model.intercept_
])
```
